# Exponential and Poisson Relationship
This notebook demonstrates how a Poisson process arises from exponentially distributed inter-arrival times. The exponential distribution models the waiting time between events, while the Poisson process counts the number of events in a fixed interval.

- For a Poisson process with rate parameter `λ`, the inter-arrival time is distributed as $\operatorname{Exp}(\lambda)$.
- The histogram below shows sampled inter-arrival times.
- The overlaid curve is the corresponding exponential probability density function $f(x)=\lambda e^{-\lambda x}$.

```python
def stimulate_poisson_process(lam=2,T=1000):
    t=0
    arrival_times=[]
    while t<T:
        wait=np.random.exponential(1/lam)

        t+=wait
        if t<T:
            arrival_times.append(t)

    return np.array(arrival_times)


        

    
```

```python
lam=2
arrival_times=stimulate_poisson_process(lam)

inter_arrivals=np.diff(np.insert(arrival_times,0,0))

plt.figure()
plt.hist(inter_arrivals,bins=40,density=True)

x=np.linspace(0,max(inter_arrivals),1000)
y=lam * np.exp(-lam * x)

plt.plot(x,y)
plt.title("Inter-arrival times (should be exponential)")
plt.xlabel("Inter-arrival time")
plt.ylabel("Density")
plt.show()
```
