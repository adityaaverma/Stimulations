```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
x_true=np.linspace(-4,4,1000)
y_true=norm.cdf(x_true)

plt.figure(figsize=(10, 6))
plt.plot(x_true, y_true, 'k-', linewidth=2.5, label='True CDF (Asli Distribution)')

sample_size=[10,50,250,500,1000]
color=['red','blue','green','pink','orange']

for n,color in zip(sample_size,color):
    data=np.random.randn(n)
    x_sorted=np.sort(data)
    y_ecdf=np.arange(1,n+1)/n
    plt.step(x_sorted, y_ecdf, where='post', color=color, alpha=0.7, 
             label=f'Empirical CDF (n={n})')

plt.title('Convergence of Empirical CDF to True CDF (Law of Large Numbers)')
plt.xlabel('Data Values (x)')
plt.ylabel('Probability F(x)')
plt.legend()
plt.grid(True)
plt.show()
```
