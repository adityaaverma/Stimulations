```python
import numpy as np
```

```python
n=100000

x = np.random.normal(0,1,n)
y = np.random.normal(0,1,n)   # make them identical

a = np.maximum(x,y)
b = np.minimum(x,y)

print("Cov(X,Y):", np.cov(x,y)[0,1])
print("Cov(max,min):", np.cov(a,b)[0,1])

print("Corr(X,Y):", np.corrcoef(x,y)[0,1])
print("Corr(max,min):", np.corrcoef(a,b)[0,1])
```

```python
import matplotlib.pyplot as plt

plt.scatter(a, b, alpha=0.2)
plt.xlabel("max(X,Y)")
plt.ylabel("min(X,Y)")
plt.title("Dependence between max and min")
plt.show()
```
