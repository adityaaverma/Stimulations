```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
A=np.random.uniform(0,1,(20,40))
B=np.random.normal(0,1,(20,40))
```

```python
_,Sa,_=np.linalg.svd(A)
_,Sb,_=np.linalg.svd(B)
```

```python
Sa
```

```python
plt.figure(figsize=(10, 6))

# Plotting singular values of A (Uniform)
plt.plot(range(1, 21), Sa, marker='o', linestyle='-', color='blue', label='Matrix A (Uniform)')

# Plotting singular values of B (Normal)
plt.plot(range(1, 21), Sb, marker='s', linestyle='-', color='red', label='Matrix B (Normal)')

# Formatting the plot
plt.title('Singular Values of 20x40 Random Matrices')
plt.xlabel('Index of Singular Value (1 to 20)')
plt.ylabel('Magnitude of Singular Value (\u03c3)')
plt.xticks(range(1, 21)) # Ensure x-axis shows integer ticks from 1 to 20
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

# Display the graph
plt.show()
```
