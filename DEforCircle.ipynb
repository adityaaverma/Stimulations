```python
import numpy as np
import plotly.graph_objects as go
```

```python
dt=0.1
n_steps=200
```

```python
y0=1.0
z0=0.0

def stimulate_forward_walk():
    y,z=y0,z0

    ys,zs=[],[]
    for _ in range(n_steps):
        ys.append(y)
        zs.append(z)

        ynew=y+dt*z
        znew=z-dt*y

        y,z=ynew,znew

    return ys,zs
```

```python
def stimulate_backward_walk():
    y,z=y0,z0
    ys,zs=[],[]

    for _ in range(n_steps):
        ys.append(y)
        zs.append(z)

        ynew=(y+dt*z)/(1+dt**2)
        znew=z-dt*ynew

        y,z=ynew,znew

    return ys,zs
```

```python
y=y0
z=z0
def stimulate_centered_walk():
    y_prev=y0
    y_curr=y0 + dt * z0
    ys=[]
    ys.append(y_prev)
    ys.append(y_curr)
    for _ in range(n_steps-2):

        y_new= (2 - dt**2)* y_curr - y_prev
        y_prev,y_curr=y_curr,y_new

        ys.append(y_new)

    return ys


        

        
```

```python
# run simulations
yf, zf = stimulate_forward_walk()
yb, zb = stimulate_backward_walk()
yc = stimulate_centered_walk()

# for centered, approximate velocity
zc = [(yc[i] - yc[i-1]) / dt for i in range(1, len(yc))]
yc = yc[1:]  # align sizes


# 🔥 Plot with Plotly
fig = go.Figure()

fig.add_trace(go.Scatter(x=yf, y=zf, mode='lines', name='Forward (spiral out)'))
fig.add_trace(go.Scatter(x=yb, y=zb, mode='lines', name='Backward (spiral in)'))
fig.add_trace(go.Scatter(x=yc, y=zc, mode='lines', name='Centered (stable)'))

fig.update_layout(
    title="Phase Space: y vs z",
    xaxis_title="y",
    yaxis_title="z",
)

fig.show(renderer="vscode")
```
